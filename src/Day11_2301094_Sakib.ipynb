```python
import numpy as np
import scipy as sp
from scipy.integrate import odeint

# Constants
w = 10.0       # UDL [N/m]
L = 5.0        # Length [m]
E = 200e9      # [Pa]
I = 1e-6       # [m^4]

def simple_beam_ode(x, u):
    u1, u2 = u     # u1 = y, u2 = dy/dx
    du1dx = u2
    du2dx = (((w*L)/2) * x - (w * x**2)/2) / (E * I)
    return [du1dx, du2dx]

def boundary_conditions(ya, yb):
    return np.array([ya[0], yb[0]])

x_values = np.linspace(0, L, 100)

# Initial guess for the solution
initial_guess = np.zeros((2, x_values.size))

# print(initial_guess)

# Solve the boundary value problem
solnSim = sp.integrate.solve_bvp(
    simple_beam_ode,
    boundary_conditions,
    x_values,
    initial_guess
)

solnSim
```

```python
# Extract the deflection values from the solution
deflection = solnSim.y[0]

# Plot the deflection
plt.figure(figsize=(8, 4.5))

plt.plot(x_values, deflection, label='Deflection')
plt.xlabel('Span of the beam (m)')
plt.ylabel('Deflection (m)')
plt.title('Deflection of Simply Supported Beam with UDL')
plt.legend()

plt.grid(True)

plt.show()
```

```python
Adist = []
Asf = []
Abm = []

def sf_bm(w, l):
    r = (w * l) / 2

    for i in np.linspace(0, l, l + 1, dtype=int):
        Adist.append(i)

        sf = r - w * i
        Asf.append(sf)

        bm = r * i - w * i * 0.5 * i
        Abm.append(bm)

    return(Adist, Asf, Abm)

result = sf_bm(50,10)
print(Adist)
print(Asf)
print(Abm)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def trapezoidal_sf_bm(w1, w2, l, num_points=200):
    """
    w1: initial load (kN/m)
    w2: final load  (kN/m)
    l:  span length (m)
    """
    x = np.linspace(0, l, num_points)

    r1 = (l / 6.0) * (2 * w1 + w2) # left support


    sf = r1 - (w1 * x) - ((w2 - w1) * (x**2)) / (2.0 * l) # SF Eqn

    bm = (r1 * x) - (w1 * (x**2) / 2.0) - ((w2 - w1) * (x**3)) / (6.0 * l) # BM Eqn

    return x, sf, bm

# input value
w1 = 20.0
w2 = 50.0
l = 10.0

x, sf, bm = trapezoidal_sf_bm(w1, w2, l)

sample_x, sample_sf, sample_bm = trapezoidal_sf_bm(w1, w2, l, num_points=int(l) + 1)
print("Distance (x):", np.round(sample_x, 2))
print("Shear Force (V):", np.round(sample_sf, 2))
print("Bending Moment (M):", np.round(sample_bm, 2))
```

```python
import numpy as np
import matplotlib.pyplot as plt

def beam_combined_analysis(l, point_loads, udls, vdls, num_points=500):
    """
    l: span (m)
    point_loads: [{'P': value (kN), 'a': distance (m)}]
    udls: [{'w': value (kN/m), 'start': (m), 'end': (m)}]
    vdls: [{'w1': ini_value, 'w2': fin_value, 'start': ini, 'end': fin}]
    """
    # Support reaction, x = 0 = R1, x = L = R2
    # moment B (x = l) -> R1 * l = Sum(Loads * distance to l)
    moment_about_b = 0.0
    total_load = 0.0

    # Point loads
    for p in point_loads:
        p_val = p['P']
        total_load += p_val
        moment_about_b += p_val * (l - p['a'])

    # UDL
    for u in udls:
        length = u['end'] - u['start']
        u_load = u['w'] * length
        cg = u['start'] + length / 2.0
        total_load += u_load
        moment_about_b += u_load * (l - cg)

    # VDL / Trapezoidal
    for v in vdls:
        length = v['end'] - v['start']
        w1, w2 = v['w1'], v['w2']
        # rect part
        u_part = min(w1, w2) * length
        u_cg = v['start'] + length / 2.0
        # trian part
        tri_part = 0.5 * abs(w2 - w1) * length
        tri_cg = v['start'] + (2.0 / 3.0 * length if w2 > w1 else 1.0 / 3.0 * length)

        v_load = u_part + tri_part
        total_load += v_load
        moment_about_b += u_part * (l - u_cg) + tri_part * (l - tri_cg)

    r1 = moment_about_b / l
    r2 = total_load - r1

    # distance grid, sf, bm
    x = np.linspace(0, l, num_points)
    sf = np.full_like(x, r1)
    bm = r1 * x

    # point load intensity
    for p in point_loads:
        dx = np.maximum(0.0, x - p['a'])
        sf -= np.where(x >= p['a'], p['P'], 0.0)
        bm -= p['P'] * dx

    # udl intensity
    for u in udls:
        dx_start = np.maximum(0.0, x - u['start'])
        dx_end = np.maximum(0.0, x - u['end'])

        # start balance load
        sf -= u['w'] * dx_start
        bm -= 0.5 * u['w'] * (dx_start ** 2)
        # extra load calcu.
        sf += u['w'] * dx_end
        bm += 0.5 * u['w'] * (dx_end ** 2)

    # VDL intensity
    for v in vdls:
        xs, xe = v['start'], v['end']
        length = xe - xs
        w1, w2 = v['w1'], v['w2']
        m = (w2 - w1) / length  # slop

        # eff. length
        xi = np.clip(x, xs, xe) - xs

        # load
        v_load = w1 * xi + 0.5 * m * (xi ** 2)
        v_moment = 0.5 * w1 * (xi ** 2) + (m / 6.0) * (xi ** 3)

        # liver arm
        after_segment = np.maximum(0.0, x - xe)

        sf -= np.where(x >= xs, v_load, 0.0)
        bm -= np.where(x >= xs, v_moment + v_load * after_segment, 0.0)

    return x, sf, bm, r1, r2

# ==================== input data====================
span = 12.0  #(m)

# point load: [{'P': load (kN), 'a': position (m)}]
point_loads = [{'P': 40.0, 'a': 3.0}]

# (UDL): [{'w': load (kN/m), 'start': , 'end': }]
udls = [{'w': 15.0, 'start': 0.0, 'end': 6.0}]

# (VDL): [{'w1': int, 'w2': fin, 'start': , 'end': }]
vdls = [{'w1': 10.0, 'w2': 30.0, 'start': 6.0, 'end': 12.0}]

# calculation
x, sf, bm, r1, r2 = beam_combined_analysis(span, point_loads, udls, vdls)

print(f"Reaction at A (R1): {r1:.2f} kN")
print(f"Reaction at B (R2): {r2:.2f} kN")
print(f"Max Bending Moment: {np.max(bm):.2f} kNm at x = {x[np.argmax(bm)]:.2f} m")

# ==================== diagram ====================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(9, 7), sharex=True)

# Shear Force Diagram (SFD)
ax1.plot(x, sf, color='crimson', lw=2)
ax1.fill_between(x, 0, sf, color='crimson', alpha=0.15)
ax1.axhline(0, color='black', lw=1)
ax1.set_title("Shear Force Diagram (SFD)", fontweight="bold")
ax1.set_ylabel("Shear Force (kN)")
ax1.grid(True, linestyle="--", alpha=0.6)

# Bending Moment Diagram (BMD)
ax2.plot(x, bm, color='navy', lw=2)
ax2.fill_between(x, 0, bm, color='navy', alpha=0.15)
ax2.axhline(0, color='black', lw=1)
ax2.set_title("Bending Moment Diagram (BMD)", fontweight="bold")
ax2.set_ylabel("Bending Moment (kNm)")
ax2.set_xlabel("Span (m)")
ax2.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
